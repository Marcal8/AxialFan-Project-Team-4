```python
import numpy as np

# --- Dades del projecte ---
Q   = 120000 / 3600      # m³/s
dp  = 300                # Pa (pressió total, Δp0)
N   = 450                # rpm
rho = 1.2                # kg/m³
mu  = 1.8e-5             # Pa·s

# --- Hipòtesis de disseny ---
D_t = 2.0                # m, diàmetre de punta
hub_ratio = 0.4          # D_h / D_t
D_h = hub_ratio * D_t

# --- Geometria i cinemàtica ---
omega = 2*np.pi*N/60
r_h, r_t = D_h/2, D_t/2
r_m = np.sqrt(0.5*(r_h**2 + r_t**2))
U   = omega*r_m
c_x = Q/(np.pi*(r_t**2 - r_h**2))

# --- Potència absorbida (η = 1) ---
P = Q*dp                 # W

# --- Euler amb entrada axial (c_theta1 = 0) i η = 1 ---
w_euler  = dp/rho               # J/kg
c_theta2 = w_euler/U            # m/s
beta1 = np.arctan(U/c_x)
beta2 = np.arctan((U - c_theta2)/c_x)   # |tan(beta2)| = |w_theta2|/c_x

print(f"D = {D_t} m, N = {N} rpm, Q = {Q:.2f} m³/s, Δp = {dp} Pa, P = {P/1e3:.2f} kW")
print(f"r_m = {r_m:.3f} m, U = {U:.2f} m/s, c_x = {c_x:.2f} m/s")
print(f"c_theta2 = {c_theta2:.2f} m/s")
print(f"beta1 = {np.degrees(beta1):.1f}°, beta2 = {np.degrees(beta2):.1f}°")

# --- Coeficients adimensionals ---
phi   = Q/(omega*D_t**3)
psi   = dp/(rho*omega**2*D_t**2)
P_hat = P/(rho*omega**3*D_t**5)
print(f"phi = {phi:.4f}, psi = {psi:.4f}, P_hat = {P_hat:.5f}, phi*psi = {phi*psi:.5f}")

Re = rho*c_x*D_t/mu
print(f"Re_D = {Re:.3e}")

# --- Velocitat i diàmetre específics ---
Ds = D_t*(dp/rho)**0.25/np.sqrt(Q)
Os = omega*np.sqrt(Q)/(dp/rho)**0.75
print(f"Ds = {Ds:.3f}, Omega_s = {Os:.3f}")
```
